```python
import coolbox
from coolbox.api import *
```

```python
wt = f"wt.hic"
hy = f"hypoxia.hic"
lt = f"latent.hic"
ko = f"napm_ko.hic"
region = "MT_H37RV_V2:1250000-1450000"
```

```python
    cool1 = DotHiC(hy,style='triangular')
    cool2 = DotHiC(ko,style='triangular')

frame = XAxis() + \
    HiCDiff(cool1, cool2, normalize='expect', diff_method='diff',style='triangular',gaussian_sigma=1.2,cmap = 'RdBu') + \
    MinValue(-2) + MaxValue(2)  + \
    BigWig('Hypoxia.bigwig',max_value=40000,min_value=0) + Color("#BC3E03") +\
    BigWig('NapM_ko.bigwig',max_value=40000,min_value=0) + Color("#BC3E03") +\
    XAxis()
frame.plot(region)
```

```python
fig=frame.plot(region)
fig.savefig("ltRv3136_down.pdf",format="pdf",dpi=500)
```
