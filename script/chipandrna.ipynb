```python
import coolbox
from coolbox.api import *
```

```python
wt = 'nahy.bedGraph'
napm = 'latent.bedgraph'
chip = 'Rv0047c_chipseq.bigwig'
```

```python
frame = XAxis() + \
    BigWig(chip, style='fill',max_value=10000,min_value=0) + Color("#1F928B") + Title("CHIP") + Spacer (height=0.2) + \
    BedGraph(wt, style='fill',max_value=100000,min_value=0) + Color("#BC3E03") + Title("WT") + Spacer (height=0.2) + \
    BedGraph(napm, style='fill',max_value=6000000,min_value=0) + Color("#BC3E03") + Title("NAPM") + Spacer (height=0.2)
```

```python
frame.plot("Chromosome:3468000-3470000")
```

```python
figure = frame.plot()
figure.savefig("Rv0217c.pdf",format="pdf",dpi=500)
```
